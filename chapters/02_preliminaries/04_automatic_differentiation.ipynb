```python
import torch

x = torch.arange(4.0)
x
```

```python
x.requires_grad_(True)
# 自己理解错了一点，x.grad 表示：输出对 x 的梯度（存放在 x 上），自己理解成 x 的梯度了！
# x.grad 指的是对 x 求导！
x.grad # 默认值是 None
```

```python
y = 2 * torch.dot(x, x)
y
```

```python
y.backward()
x.grad
```

```python
x.grad == 4 * x
```

```python
# 默认情况下，当 x 参与计算时，会累积梯度，在下一次不希望累积的反向传播前，需要清除其梯度
x.grad.zero_() # type: ignore
y = x.sum()
y.backward()
x.grad
```

```python
x.grad.zero_()
y = x * x
y.sum().backward()
x.grad
```

```python
x = torch.arange(4.0)
x.requires_grad_(True)
# 此时 x.grad 还是 None
# x.grad.zero_()
y = x * x
# 不传参数调用 backward() 时，输出必须只有一个元素。这里 y 有四个元素，需要先求和，或者显式传入上游梯度。
# y.backward()
# x.grad
torch.autograd.functional.jacobian(lambda x:x*x,y)
```
